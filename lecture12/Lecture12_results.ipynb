# Exercise 12.1

## Results


I trained the same DNN for 15 epochs with four optimisers (batch size 64, default hyper-parameters). The MNIST test set is used as validation data, so the final numbers are slightly optimistic.

| Optimizer | Train acc. | Val. acc. | Train loss | Val. loss | 
|---|---|---|---|---|
| SGD | 0.9496 | 0.9624 | 0.1781 | 0.1262 |
| Adam | 0.9939 | 0.9828 | 0.0197 | 0.0819 | 
| RMSprop | 0.9911 | 0.9813 | 0.0360 | 0.1393 | 
| Adamax | 0.9900 | 0.9815 | 0.0348 | 0.0622 | 

**Comments.**
* **SGD**  is the slowest: accuracy and loss are still improving at epoch 15, so more epochs (or momentum / a larger learning rate) are needed. The training accuracy is below the validation one because dropout is active only during training.
* **Adam and RMSprop** converge in a few epochs, but after epoch 3-4 the validation loss increases while the training loss keeps decreasing: the network starts to overfit. 
* **Adamax** is the most stable: train and validation curves stay close, and the validation loss (0.062) is the lowest and still decreasing Since the best epoch differs among optimisers, an early-stopping criterion on the validation loss is the sensible way to choose the number of epochs.

# Exercise 12.2

## Results


The CNN is: Conv2D(10, 5×5) → MaxPooling(2×2) → Conv2D(20, 5×5) → MaxPooling(2×2) → Dropout(0.25) → Flatten → Dense(128) → Dropout(0.5) → softmax(10), with **47 658** parameters. I trained it for 15 epochs (batch size 64) with two optimizers:

| Optimizer | Test accuracy | Test loss |
|---|---|---|
| Adam | 0.9930 | 0.0231 |
| RMSprop | 0.9935 | 0.0208 |

**Comments.**
* The CNN has about 7.5 times fewer parameters (47 658 vs 355 110) and still performs better: local filters shared across the image exploit the spatial structure of the digits, which a fully connected network ignores.
* The difference between Adam and RMSprop is not significant for a single run, so no optimizer can be declared better.
* The validation accuracy is higher than the training accuracy because dropout is active only during training. Validation loss is still decreasing at epoch 15, so the network is not overfitting.
